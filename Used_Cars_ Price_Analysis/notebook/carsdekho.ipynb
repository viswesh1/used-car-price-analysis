```python
import pandas as pd
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
```

```python
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

print("Browser opened successfully")
```

```python
url = "https://www.cardekho.com/used-cars+in+hyderabad"
driver.get(url)
time.sleep(5)
print("page loaded successfully")
```

```python
soup = BeautifulSoup(driver.page_source, "html.parser")
```

```python
cards = soup.find_all("div", class_="NewUcExCard")

print('total car cards found:',len(cards))
```

```python
brands = ["Maruti", "Hyundai", "Honda", "Tata", "Mahindra", "Toyota",
          "Ford", "Volkswagen", "Skoda", "BMW", "Mercedes-Benz",
          "Audi", "Kia", "Renault", "Nissan", "Jeep", "MG", "BYD",
          "Land Rover","Citroen","Volvo","Mini Cooper","Jaguar","Lexus",
          "Porsche","Chevrolet","Isuzu","Datsun","Rolls-Royce","Mitsubishi",
          "Lamborghini "]
```

```python
all_cars = []
```

```python
for card in cards:
    try:
        #--Title--
        title = card.find("h3",class_="title").get_text(strip=True)
        
        #--year--
        year = None
        for y in range(1990,2027):
            if str(y) in title:
                year = str(y)
                title = title.replace(str(y),str(y) + " ")
                break

        #--car name--
        car_name = title.strip()

        if year:
                car_name = car_name.replace(year, "").strip()
        
        #--Brand--
        car_brand =  None
        for b in brands:
            if b in title:
                car_brand = b
                break

        #--Details--
        details = card.find("div", class_="dotsDetails")
        if details:
            parts = [x.strip() for x in details.get_text(strip=True).split("•")]
            km_driven = parts[0] if len(parts) > 0 else None
            fuel = parts[1] if len(parts) > 1 else None
            transmission = parts[2] if len(parts) > 2 else None
        else:
            km_driven = fuel = transmission = None

        #--price--
        price_tag = card.find("div", class_="priceAssured")
        price = price_tag.get_text(strip=True) if price_tag else None

        #---Append---
        all_cars.append({
            "car_brand": car_brand,
            "car_name":car_name,
            "year":year,
            "km_driven":km_driven,
            "fuel":fuel,
            "transmission":transmission,
            "price":price,
            "city":"Hyderabad"
        })

    except Exception:
        continue

        
        

        
```
